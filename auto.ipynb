```python
#input fixed
x = torch.tensor(2.0)
#learning no.
w = torch.tensor(3.0, requires_grad=True)
#correct answer
y_true = torch.tensor(10.0)
#learning rate
lr = 0.1
y_pred = w*x
loss = (y_pred - y_true) **2
loss.backward()
print("Before update")
print("w:",w.item())
print("loss",loss.item())
print("gradient",w.grad.item())
with torch.no_grad():
    w -= lr * w.grad
w.grad.zero_()
print("After update")
print("w",w.item())
```
